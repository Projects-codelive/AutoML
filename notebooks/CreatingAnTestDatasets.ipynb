```python
import pandas as pd

# ==========================================
# 1. Regression Dataset -> Save as .csv
# ==========================================
regression_data = {
    'Area_sqft': [1200, 2500, 850, 1800, 3200, 1500],
    'Bedrooms': [2, 4, 1, 3, 5, 3],
    'Age_years': [15, 5, 40, 10, 2, 20],
    'Distance_to_City_miles': [12, 5, 2, 15, 8, 20],
    'Price_Thousands': [250, 620, 180, 315, 850, 210] # Continuous Target
}
df_regression = pd.DataFrame(regression_data)

# Save to CSV (index=False prevents pandas from writing row numbers to the file)
df_regression.to_csv('../data/regression_sample.csv', index=False)
print("Saved regression_sample.csv")


# ==========================================
# 2. Classification Dataset -> Save as .xlsx
# ==========================================
classification_data = {
    'Credit_Score': [750, 580, 620, 490, 810, 680],
    'Income': [85000, 42000, 55000, 38000, 120000, 60000],
    'Debt_to_Income_Ratio': [0.25, 0.65, 0.40, 0.70, 0.15, 0.35],
    'Late_Payments_Past_Year': [0, 3, 1, 5, 0, 0],
    'Defaulted': ['No', 'Yes', 'No', 'Yes', 'No', 'No'] # Categorical Target
}
df_classification = pd.DataFrame(classification_data)

# Save to Excel (Make sure you have 'openpyxl' installed)
df_classification.to_excel('../data/classification_sample.xlsx', index=False)
print("Saved classification_sample.xlsx")


# ==========================================
# 3. Clustering Dataset -> Save as .json
# ==========================================
clustering_data = {
    'Customer_Age': [25, 45, 22, 50, 35, 28],
    'Annual_Income_k': [30, 120, 25, 110, 65, 35],
    'Spending_Score_1_to_100': [85, 15, 90, 10, 50, 80],
    'Time_on_App_mins': [45, 10, 55, 5, 20, 40]
    # No Target Column!
}
df_clustering = pd.DataFrame(clustering_data)

# Save to JSON (orient='records' formats it beautifully as a list of dictionaries)
df_clustering.to_json('../data/clustering_sample.json', orient='records', indent=4)
print("Saved clustering_sample.json")
```
